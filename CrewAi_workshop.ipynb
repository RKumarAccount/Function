```python
import os
from crewai import Agent, Task, Crew, LLM
from crewai_tools import SerperDevTool
```

```python
os.environ['SERPER_API_KEY'] = 'use'
#os.environ['OPENAI_API_KEY'] = ''
os.environ['GEMINI_API_KEY'] = 'use'
```

```python
search_tool = SerperDevTool()

llm=LLM(
    model='gemini/gemini-2.5-flash',
    api_key=os.environ['GEMINI_API_KEY'],
    temperature=0.5,
)

teacher=Agent(
    role="Teacher",
    goal='Explain the concept clearly step by step',
    llm=llm,
    verbose=True,
    backstory='Experienced Teacher who explian the with expample'
)

researcher=Agent(
    role="Researcher",
    goal='find latest information',
    llm=llm,
    tools=[search_tool],
    verbose=True,
    backstory='expert reaching in latest information'

)

simplifier=Agent(
    role="Simplifier",
    goal='Making things easy',
    llm=llm,
    verbose=True,
    backstory='breack complex ideas into simple languae'
)

student=Agent(
    role="Student",
    goal='take notes',
    backstory='write simple notes like learner',
    llm=llm,
    verbose=True
)

examiner=Agent(
    role="Examiner",
    goal='create quetsions',
    llm=llm,
    verbose=True,
    backstory='test unsrstanding'
)

topic='what is diffrence between ai agent vs agentic ai vs mcp vs generative ai'


task1=Task(
    description=f'explain  {topic} in simple steps with example',
    expected_output='step-by-step by explanation',
    agent=teacher,
)

task2=Task(
    description=f'search find and import 3 points about {topic}',
    expected_output='3 clear points',
    agent=researcher
)

task3=Task(
    description=f'simplify the explanation of {topic}',
    expected_output='very simple explanation',
    agent=simplifier,
)

task4=Task(
    description=f'write short notes on {topic} like a student',
    expected_output='simple notes',
    agent=student,
)

task5=Task(
    description=f'create 5 questions about {topic}',
    expected_output='step by step explanation',
    agent=examiner,
)

crew = Crew(
    agents=[teacher, researcher, simplifier, student, examiner],
    tasks=[task1, task2, task3, task4, task5],
    verbose=True
)

result=crew.kickoff()

print(result)

```
